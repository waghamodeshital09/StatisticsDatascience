```python
#Q1.Import all important libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
```

```python
#Q2. Generate 1,000 random values using NumPy.
data = np.random.rand(1000)
print(data[:10])
```

```python
#Q3. Plot a histogram.
sns.histplot(data, kde=True)
plt.show()
```

```python
#Q4. Generate 1,000 values using Normal Distribution.
data = np.random.randn(1000)
sns.histplot(data, kde=True)
plt.show()
```

```python
#Level 2 – Log Transformation and Box-Cox
```

```python
#Q5. Apply Log Transformation to Pareto data.
data = stats.pareto.rvs(b=2.62, size=1000)
log_data = np.log(data)
sns.histplot(log_data, kde=True)
plt.title("Log Transformation")
plt.show()
```

```python
#Q6. Apply Box-Cox transformation.
data = stats.pareto.rvs(b=2.62, size=1000)
transformed_data, lam = stats.boxcox(data)
print("Lambda:", lam)
print("Transformed Data:", transformed_data[:10])
```

```python
#Q7.Compare original and transformed data.
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
sns.histplot(data, kde=True)
plt.title("Original Data")
plt.subplot(1,2,2)
sns.histplot(transformed_data, kde=True)
plt.title("Box-Cox Data")
plt.show()#We compare both distributions to see whether Box-Cox reduces skewness
```

```python
#Q8.Calculate skewness before and after Box-Cox.
print("Original:", stats.skew(data))
print("After Box-Cox:", stats.skew(transformed_data))
```

```python
#Q9.Apply Box-Cox to a manually created dataset.
data = [10, 20, 30, 40, 50, 100, 150, 200]
transformed_data, lam = stats.boxcox(data)
print(transformed_data)
print("Lambda:", lam)
```

```python
#Level 3 – Other Non-Gaussian distributions
```

```python
#Q10. Generate Exponential distribution.
data = stats.expon.rvs(scale=2,size=1000)
sns.histplot(data, kde=True)
plt.title("Exponential Distribution")
plt.show()
```

```python
#Q11. Generate Weibull distribution.
data = stats.weibull_min.rvs(c=2,size=1000)
sns.histplot(data, kde=True)
plt.title("Weibull Distribution")
plt.show()#c = Shape parameter.
           #Weibull distribution is used in reliability and lifetime analysis.
```

```python
#Q12.Generate Gamma distribution.
data = stats.gamma.rvs(a=2,size=1000)
sns.histplot(data, kde=True)
plt.title("Gamma Distribution")
plt.show()#Gamma distribution is a continuous distribution for positive values.
```

```python
#Q13. Generate Logistic distribution.
data = stats.logistic.rvs(loc=0,scale=1,size=1000)
sns.histplot(data, kde=True)
plt.title("Logistic Distribution")
plt.show()
```

```python
#Q14. Compare all four distributions.
pareto_data = stats.pareto.rvs(b=2.62, size=1000)
expon_data = stats.expon.rvs(size=1000)
weibull_data = stats.weibull_min.rvs(c=2, size=1000)
gamma_data = stats.gamma.rvs(a=2, size=1000)

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

sns.histplot(pareto_data, kde=True, ax=axes[0,0])
axes[0,0].set_title("Pareto")

sns.histplot(expon_data, kde=True, ax=axes[0,1])
axes[0,1].set_title("Exponential")

sns.histplot(weibull_data, kde=True, ax=axes[1,0])
axes[1,0].set_title("Weibull")

sns.histplot(gamma_data, kde=True, ax=axes[1,1])
axes[1,1].set_title("Gamma")

plt.tight_layout()
plt.show()#This helps you understand how different non-Gaussian distributions look.
```

```python
#Level 5 – Fit Distribution and CDF
```

```python
#Q19. Fit Weibull distribution to rainfall data.
rainfall = np.array([5, 8, 10, 12, 15, 17, 20, 22, 25, 30])
params = stats.weibull_min.fit(rainfall)
print(params)
```

```python
#Q20.Calculate CDF.
cdf = stats.weibull_min.cdf(20, *params)
print(cdf)
```

```python
#Q21.Calculate probability of rainfall greater than or equal to 20 cm.
probability = 1 - cdf
print(probability)
```
