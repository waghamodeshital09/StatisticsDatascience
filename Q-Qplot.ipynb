# Topic 4: Q-Q Plot (Quantile-Quantile Plot)

```python
# Import libraries
import numpy as np
import pandas as pd
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Generate normal distribution data
data=np.random.normal(0,1,1000)
```

```python
data
```

```python
#Create Q-Q plot
stats.probplot(data,dist="norm",plot=plt)
plt.title("Q-Q plot")
plt.show()
```

```python
#Create CLT function
def CLT(data,no_of_samples,sample_size):
    result=[]
    for i in range(no_of_samples):
        samples=np.random.choice(data,size=sample_size,replace=True)
        result.append(np.mean(samples))
    return np.array(result)   
```

```python
#Apply CLT function
data2=df["Price"]
no_of_samples=1000
sample_size=30
result_values=CLT(data2,no_of_samples,sample_size)
```

```python
#Display sample means
result_values
```

```python
#Plot sample means
sns.distplot(result_values)
```

```python
#Create Q-Q plot
stats.probplot(result_values,dist=stats.uniform,plot=plt)
plt.title("Q-Q plot")
plt.show()
```
