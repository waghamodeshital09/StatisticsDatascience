```python
#Q16.Create a function to calculate mean of discrete uniform distribution.
def discrete_mean(a, b):
    mean = (a+b)/2
    return mean
print(discrete_mean(1, 6))
print(discrete_mean(1, 10))
```

```python
#Q17.Create a function to calculate variance of discrete uniform distribution.
def discrete_variance(a, b):
    n = b-a+1
    variance = (n**2-1)/12
    return variance
print(discrete_variance(1, 6))
```

```python
#Q18. Create a function to calculate PDF of continuous uniform distribution.
def continuous_pdf(a, b):
    pdf = 1/(b-a)
    return pdf
print(continuous_pdf(0, 10))
print(continuous_pdf(5, 15))
```

```python
#Q19. Generate 10,000 uniform random numbers and plot a histogram.
import matplotlib.pyplot as plt
data = np.random.uniform(0, 10, 10000)
plt.hist(data, bins=20)
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Uniform Distribution")
plt.show()
```

```python
#Q20.Generate uniform random data and create a Q-Q plot.
import scipy.stats as stats
import matplotlib.pyplot as plt
data = np.random.uniform(0, 10, 1000)
stats.probplot(
    data,
    dist=stats.uniform,
    sparams=(0, 10),
    plot=plt
)
plt.title("Uniform Distribution Q-Q Plot")
plt.show()
```
