# Permutation Test

```python
#Q1.Create sample1 using NumPy.
import numpy as np
sample1=np.array([10,20,30])
print(sample1)
```

```python
#Q2.Create sample2 using NumPy.
sample2=np.array([15,25,35])
print(sample2)
```

```python
#Q3.Find the mean of both samples.
mean1=np.mean(sample1)
mean2=np.mean(sample2)
print(mean1)
print(mean2)
```

```python
#Q4.Find the difference between two means.
original_diff=abs(mean1-mean2)
print(original_diff)
```

```python
#Q5.Combine both samples.
combined_sample=np.concatenate((sample1,sample2))
print(combined_sample)
```

```python
#Q6.Shuffle the combined sample.
shuffled_data=np.random.permutation(combined_sample)
print(shuffled_data)
```

```python
#Q7.Divide shuffled data into two samples.
new_sample1=shuffled_data[:3]
new_sample2=shuffled_data[3:]
print(new_sample1)
print(new_sample2)
```

```python
#Q8.Find the mean of both shuffled samples.
new_mean1=np.mean(new_sample1)
new_mean2=np.mean(new_sample2)
print(new_mean1)
print(new_mean2)
```

```python
#Q9.Calculate the difference between new means.
new_diff=abs(new_mean1-new_mean2)
print(new_diff)
```

```python
#Q10.Print original and shuffled differences.
print("Original Difference:",original_diff)
print("New Difference:",new_diff)
```

```python
#Q11.Perform one permutation test.#We combine,shuffle,divide and calculate the new difference once.
combined_sample=np.concatenate((sample1,sample2))
shuffled_data=np.random.permutation(combined_sample)

new_sample1=shuffled_data[:len(sample1)]
new_sample2=shuffled_data[len(sample1):]

new_diff=abs(np.mean(new_sample1)-np.mean(new_sample2))
print(new_diff)
```

```python
#Q12.Perform 100 permutations and store differences.
diff_lst=[]
for i in range(100):
    shuffled_data=np.random.permutation(combined_sample)
    
    new_sample1=shuffled_data[:len(sample1)]
    new_sample2=shuffled_data[len(sample1):]

    new_diff=abs(np.mean(new_sample1)-np.mean(new_sample2))
    diff_lst.append(new_diff)

    
print(diff_lst)    
    
```

```python
#Q13.Convert difference list into NumPy array.
diff_arr=np.array(diff_lst)
print(diff_arr)
```

```python
#Q14.Count differences greater than or equal to original difference.
count=np.sum(diff_arr>=original_diff)
print(count)
```

```python
#Q15.Calculate p-value.
p_value=count/100
print(p_value)
```
