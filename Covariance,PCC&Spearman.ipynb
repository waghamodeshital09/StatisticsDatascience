```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    'Height': [168, 174, 180, 184, 190, 155],
    'Weight': [60, 58, 74, 55, 80, 65],
    'Salary': [20000, 25000, 30000, 35000, 40000, 22000],
    'Experience': [1, 2, 3, 4, 5, 1]
})

print(df)
```

```python
#Q1. Create two NumPy arrays X and Y.
import numpy as np
x = np.array([10, 20, 30, 40, 50])
y = np.array([15, 25, 35, 45, 55])
print(x)
print(y)
```

```python
#Q2. Calculate mean of X and Y.
print(np.mean(x))
print(np.mean(y))
```

```python
#Q3. Calculate variance of X and Y.
print(np.var(x))
print(np.var(y))
```

```python
#Q5. Calculate covariance using NumPy.
cov = np.cov(x, y, ddof=0)
print(cov[0, 1])
```

```python
#Q6. Calculate Pearson Correlation using NumPy.
pcc = np.corrcoef(x, y)
print(pcc[0, 1])
```

```python
#Q7. Calculate Spearman Rank Correlation using SciPy.
from scipy.stats import spearmanr
result = spearmanr(x, y)
print(result.statistic)
```

```python
#Q8. Find ranks of X and Y using Pandas.
import pandas as pd
df['Rank_X'] = df['Height'].rank()
df['Rank_Y'] = df['Weight'].rank()
print(df)
```

```python
#Q9. Calculate Pearson correlation using Pandas.
pcc = df['Height'].corr(df['Weight'])
print(pcc)
```

```python
#Q10. Calculate Spearman correlation using Pandas.
spearman = df['Height'].corr( df['Weight'], method='spearman')
print(spearman)
```

```python
#Q11. Calculate all three methods for the same dataset.
cov = df['Height'].cov(df['Weight'])
pcc = df['Height'].corr(df['Weight'])
spearman = df['Height'].corr(df['Weight'],method='spearman')
print("Covariance:", cov)
print("PCC:", pcc)
print("Spearman:", spearman)
```

```python
#Q17. Create a non-linear monotonic dataset and compare PCC with Spearman.
from scipy.stats import spearmanr
x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]
pcc = np.corrcoef(x, y)[0, 1]
spearman = spearmanr(x, y).statistic
print("PCC:", pcc)
print("Spearman:", spearman)
```

```python
#Q18. Create a non-linear non-monotonic dataset and compare PCC with Spearman.
x = [-3, -2, -1, 0, 1, 2, 3]
y = [9, 4, 1, 0, 1, 4, 9]
pcc = np.corrcoef(x, y)[0, 1]
spearman = spearmanr(x, y).statistic
print("PCC:", pcc)
print("Spearman:", spearman)
```

```python
#Q19. Add an outlier and compare Pearson and Spearman.
x = [1, 2, 3, 4, 5, 6]
y = [2, 4, 6, 8, 10, 100]
pcc = np.corrcoef(x, y)[0, 1]
spearman = spearmanr(x, y).statistic
print("PCC:", pcc)
print("Spearman:", spearman)
```
