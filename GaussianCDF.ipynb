# Gaussian CDF(cumulative distribution function)

```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df=pd.read_csv("laptop_data.csv")
print(df.shape)
```

```python
sns.displot(df["Price"],kde=True)
```

```python
import math
```

```python
#Create a Gaussian CDF function
def Gaussian_CDF(data):
    result=[]
    meu=np.mean(data)
    sd=np.std(data)
    for i in data:
        a1=(i-meu)/(np.sqrt(2)*sd)
        ans=1/2*(1+math.erf(a1))
        result.append(ans)
    return np.array(result)    
```

```python
#Sort laptop prices
data=np.sort(df["Price"])
```

```python
#Call the CDF function
cdf_values = Gaussian_CDF(data)
```

```python
#Display CDF values
cdf_values
```

```python
#Import plotting library
import matplotlib.pyplot as plt
```

```python
#Set graph labels
plt.xlabel("x_values")
plt.ylabel("CDF values")
plt.title("CDF")
```

```python
#Full practice code
#Laptop price CDF — complete program
import numpy as np
import pandas as pd
import seaborn as sns
import math
import matplotlib.pyplot as plt
df=pd.read_csv("laptop_data.csv")
def Gaussian_CDF(data):
    result=[]
    meu=np.mean(data)
    sd=np.mean(data)
    for i in data:
        a1=(i-meu)/(np.sqrt(2)*sd)
        ans=1/2*(1+math.erf(a1))
        result.append(ans)
    return np.array(result)
data=np.sort(df["Price"])
cdf_values=Gaussian_CDF(data)
plt.xlabel("x_values")
plt.ylabel("CDF values")
plt.title("CDF")
plt.plot(data,cdf_values)
plt.show()
```
