# Topic 3: Chebyshev’s Inequality

```python
import numpy as np 
import pandas as pd
import seaborn as sns
```

```python
#Create data
data = np.random.uniform(0,1,100)
```

```python
#Calculate the mean
mean=np.mean(data)
print(mean)
```

```python
#Calculate standard deviation
sd = np.std(data)
print(sd)
```

```python
#Build the Chebyshev function
def chebyshevs_inequality(data,k):
    mean=np.mean(data)
    sd=np.std(data)
    LL=mean-k*sd
    UL=mean+k*sd
    per=(1-(1/k**2))*100
    return(LL,UL),per
```

```python
#Set the value of k
k=2
print(k)
```

```python
#Calculate the percentage
k=2
per=(1-(1/k**2))*100
print(per)
```

```python
#Call the function
data=np.random.uniform(0,1,100)
k=2
chebyshevs_inequality(data,k)
```

```python
#Store the results
result=chebyshev_inequality(data,k)
print(result)
```

```python
#Display the limits separately
limits,percentage=chebyshev_inequality(data,k)
print("Limits:",limits)
print("percentage:",percentage)
```

# Central Limit Theorem (CLT)

```python
#Create the CLT function
def central_limit_therom(data,no_of_samples,sample_size):
    result=[]
    for i in range(no_of_samples):
        sample = np.random.choice(data,size=sample_size,replace=True)
        result.append(np.mean(sample))
    data=np.array(result)
    return data 
```

```python
data=np.random.randint(1,1000,1000)
no_of_samples = 1000
sample_size = 30
data = central_limit_therom(data,no_of_samples,sample_size)
data
```
