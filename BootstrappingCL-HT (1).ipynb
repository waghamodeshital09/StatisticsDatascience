```python
#Q14. Find 90% CI for standard deviation.
std_values = []
for i in range(1000):
    sample = np.random.choice(data, 5, replace=True)
    std_values.append(np.std(sample, ddof=1))
ll = np.percentile(std_values, 5)
ul = np.percentile(std_values, 95)
print("90% CI:", ll, ul)
```

```python
#Q16.Create two arrays representing heights of students.
class1 = np.array([174, 158, 160, 165, 170])
class2 = np.array([168, 192, 160, 175, 180])
print(class1)
print(class2)
```

```python
#Q17.Calculate average height of both classes.
print(np.mean(class1))
print(np.mean(class2))
```

```python
#Q18. Find difference between both average heights.
mean1 = np.mean(class1)
mean2 = np.mean(class2)
print(mean1 - mean2)
```

```python
#Q19.Perform independent two-sample t-test.
from scipy.stats import ttest_ind
result = ttest_ind(class1, class2, equal_var=False)
print(result)
```

```python
#Q20.Interpret the p-value.
p_value = result.pvalue
if p_value < 0.05:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

```python
#For these questions, first create a reusable bootstrap function
def bootstrap_ci(data, statistic, confidence=95,
                 k=1000, sample_size=None):

    if sample_size is None:
        sample_size = len(data)

    values = []

    for i in range(k):
        sample = np.random.choice(
            data, sample_size, replace=True
        )

        values.append(statistic(sample))

    alpha = 100 - confidence

    ll = np.percentile(values, alpha / 2)
    ul = np.percentile(values, 100 - alpha / 2)

    return ll, ul
```

```python
#Q21.Create a function to calculate median CI.
data = np.array([10, 20, 30, 40, 50, 60, 70, 80])
result = bootstrap_ci(data, np.median, 95)
print(result)
```

```python
#Q22. Calculate mean CI.
result = bootstrap_ci(data, np.mean, 95)
print(result)
```

```python
#Q23. Calculate SD CI.
result = bootstrap_ci(
    data,
    lambda x: np.std(x, ddof=1),
    90
)
print(result)
```

```python
#Q24. Calculate 90th percentile CI.
result = bootstrap_ci(
    data,
    lambda x: np.percentile(x, 90),
    95
)
print(result)
```

```python
#Q25.Compare original sample median with bootstrap median CI.
original_median = np.median(data)
ll, ul = bootstrap_ci(data, np.median, 95)
print("Original Median:", original_median)
print("Lower Limit:", ll)
print("Upper Limit:", ul)
```
